# CineBook Big-Data Analytics & Insights 🎬

This notebook connects to the CineBook MongoDB big-data layer, pulls analytical pipelines, loads them into **pandas DataFrames**, and generates visualizations for:
1. Box office gross revenue and ticket velocity
2. Peak-hour screening heatmap
3. Payment instrument market share (UPI dominant)
4. Audience taste profiles by genre and language
5. Customer lifetime value and segmentation

```python
import os
from pymongo import MongoClient
import pandas as pd
from datetime import datetime

# Connect to MongoDB (local or Atlas)
mongo_uri = os.getenv("MONGO_URI", "mongodb://localhost:27017")
db_name = os.getenv("MONGO_DB_NAME", "cinebook_analytics")
client = MongoClient(mongo_uri)
db = client[db_name]

print(f"Connected to {db.name}")
print(f"Total Booking Events: {db.booking_events.estimated_document_count():,}")
print(f"Total Page Views:     {db.page_view_events.estimated_document_count():,}")
```

```python
from app.mongo.analytics_queries import get_top_movies, get_payment_method_distribution, get_customer_segmentation

# 1. Top Movies by Revenue
top_movies = get_top_movies(db, limit=10)
df_movies = pd.DataFrame(top_movies)
print("Top Blockbusters:")
df_movies
```

```python
# 2. Payment Method Distribution
payments = get_payment_method_distribution(db)
df_payments = pd.DataFrame(payments)
df_payments['share_pct'] = (df_payments['total_revenue'] / df_payments['total_revenue'].sum()) * 100
df_payments
```

```python
# 3. Customer Segments
seg_data = get_customer_segmentation(db)
df_segments = pd.DataFrame(seg_data["segments"])
print(f"Total Unique Customers: {seg_data['total_unique_customers']:,} | Repeat Rate: {seg_data['repeat_rate']}%")
df_segments
```
